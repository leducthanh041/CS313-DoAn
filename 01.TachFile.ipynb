```python
import pandas as pd
from pathlib import Path

# =========================
# 1) Đọc file tổng
# =========================
in_path = Path(r"./data_thuydien_preprocessed/du_lieu_thuy_dien_processed.csv")
if not in_path.exists():
    raise FileNotFoundError(f"Không thấy file: {in_path.as_posix()}")

df = pd.read_csv(in_path)

# =========================
# 2) Kiểm tra cột ten_ho
# =========================
if "ten_ho" not in df.columns:
    raise KeyError("Không tìm thấy cột 'ten_ho' trong file.")

# (tuỳ chọn) xem các giá trị hiện có
print("Các giá trị ten_ho:", df["ten_ho"].dropna().unique())

# =========================
# 3) Tách và lưu 3 file riêng
# =========================
out_dir = Path("./data_thuydien_preprocessed/splitted_by_dam")
out_dir.mkdir(parents=True, exist_ok=True)

dams = ["Sông Ba Hạ", "Hòa Bình", "Trị An"]

for dam in dams:
    df_dam = df[df["ten_ho"].astype(str).str.strip() == dam].copy()

    # đặt tên file an toàn
    safe_name = (
        dam.replace(" ", "_")
           .replace("à", "a").replace("á", "a").replace("ạ", "a").replace("ả", "a").replace("ã", "a")
           .replace("â", "a").replace("ầ", "a").replace("ấ", "a").replace("ậ", "a").replace("ẩ", "a").replace("ẫ", "a")
           .replace("ă", "a").replace("ằ", "a").replace("ắ", "a").replace("ặ", "a").replace("ẳ", "a").replace("ẵ", "a")
           .replace("ò", "o").replace("ó", "o").replace("ọ", "o").replace("ỏ", "o").replace("õ", "o")
           .replace("ô", "o").replace("ồ", "o").replace("ố", "o").replace("ộ", "o").replace("ổ", "o").replace("ỗ", "o")
           .replace("ơ", "o").replace("ờ", "o").replace("ớ", "o").replace("ợ", "o").replace("ở", "o").replace("ỡ", "o")
           .replace("ù", "u").replace("ú", "u").replace("ụ", "u").replace("ủ", "u").replace("ũ", "u")
           .replace("ư", "u").replace("ừ", "u").replace("ứ", "u").replace("ự", "u").replace("ử", "u").replace("ữ", "u")
           .replace("ì", "i").replace("í", "i").replace("ị", "i").replace("ỉ", "i").replace("ĩ", "i")
           .replace("đ", "d")
    )

    out_path = out_dir / f"du_lieu_thuy_dien_processed_{safe_name}.csv"
    df_dam.to_csv(out_path, index=False, encoding="utf-8")

    print(f"{dam}: {len(df_dam):,} rows -> {out_path.as_posix()}")

print("DONE.")
```
