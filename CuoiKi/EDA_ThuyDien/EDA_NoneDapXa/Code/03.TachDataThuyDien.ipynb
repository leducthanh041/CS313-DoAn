```python
import pandas as pd
from pathlib import Path
```

```python
path = "../../.././data_thuydien/data_thuydien_noneVietnamese.csv"

df = pd.read_csv(path)
```

```python
df["ten_ho"].unique()
```

```python
output_dir = Path("../../.././data_thuydien/dataCacDap")
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
dams = ["Sông Ba Hạ", "Hòa Bình", "Trị An"]

for dam in dams:
    df_dam = df[df["ten_ho"] == dam].copy()
    
    file_name = f"data_thuydien_{dam.replace(' ', '_')}.csv"
    file_path = output_dir / file_name
    
    df_dam.to_csv(file_path, index=False)
    
    print(f"{dam}: {df_dam.shape[0]} rows -> saved to {file_path}")
```
